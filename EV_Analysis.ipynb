```python
import glob
import os
from IPython.display import Image, display
import pandas as pd

# Run the main project pipeline
!python main.py
```

```python
# Automatically display every chart saved in outputs/graphs
graph_files = sorted(glob.glob("outputs/graphs/*.*"))

for file in graph_files:
  print(f"Displaying: {os.path.basename(file)}")
  if file.endswith((".png", ".jpg", ".jpeg")):
    display(Image(filename=file))
```

```python
# Display any CSV analysis tables
for csv_file in sorted(glob.glob("outputs/analysis/*.csv")):
  print(f"\n--- {os.path.basename(csv_file)} ---")
  display(pd.read_csv(csv_file).head(10))

# Display text/markdown reports
for rep_file in sorted(glob.glob("outputs/reports/*.*")):
  print(f"\n--- {os.path.basename(rep_file)} ---")
  with open(rep_file, "r", encoding="utf-8", errors="ignore") as f:
    print(f.read())
```
